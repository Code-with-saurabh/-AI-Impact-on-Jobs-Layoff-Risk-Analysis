```python
import sys
import os
sys.path.insert(0, os.path.dirname(os.path.dirname(os.path.abspath('__file__'))))

import pandas as pd
import numpy as np

import config
from src.data_loading import load_dataset
from src.data_cleaning import handle_missing_values, remove_duplicates, standardize_column_names
from src.feature_engineering import run_feature_engineering
from src.preprocessing import (
    encode_categorical_columns, scale_numerical_columns,
    split_features_target, train_test_split_data
)

print('All imports successful!')
```

## 1. Load and Clean Data

```python
df = load_dataset(config.RAW_DATASET_PATH)
print(f'Original shape: {df.shape}')

df = handle_missing_values(df, strategy='drop')
df = remove_duplicates(df)
df = standardize_column_names(df)

print(f'Cleaned shape: {df.shape}')
df.head()
```

## 2. Feature Engineering

```python
df_engineered = run_feature_engineering(df)
print(f'\nShape after feature engineering: {df_engineered.shape}')
print(f'\nNew columns added:')
new_cols = set(df_engineered.columns) - set(df.columns)
for col in sorted(new_cols):
    print(f'  - {col}')
```

```python
df_engineered.head()
```

## 3. Feature Engineering Details

```python
print('=== Experience Categories ===')
print(df_engineered['experience_category'].value_counts())

print('\n=== Age Groups ===')
print(df_engineered['age_group'].value_counts())

print('\n=== AI Intensity Score ===')
print(f'Min: {df_engineered["ai_intensity_score"].min():.1f}')
print(f'Max: {df_engineered["ai_intensity_score"].max():.1f}')
print(f'Mean: {df_engineered["ai_intensity_score"].mean():.1f}')
```

## 4. Encode Categorical Columns

```python
df_encoded, encoders = encode_categorical_columns(df_engineered, config.TARGET_COLUMN)

print('\nEncoded columns:')
for col, encoder in encoders.items():
    print(f'  {col}: {list(encoder.classes_)}')
```

## 5. Split Features and Target

```python
X, y = split_features_target(df_encoded, config.TARGET_COLUMN)
print(f'\nFeatures shape: {X.shape}')
print(f'Target shape: {y.shape}')
```

## 6. Scale Numerical Columns

```python
X_scaled, scaler = scale_numerical_columns(X)
print(f'\nScaled features shape: {X_scaled.shape}')
X_scaled.head()
```

## 7. Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split_data(
    X_scaled, y, test_size=config.TEST_SIZE, random_state=config.RANDOM_STATE
)

print(f'\nTraining set: {X_train.shape}')
print(f'Testing set: {X_test.shape}')
```

## 8. Target Distribution in Train/Test

```python
print('=== Train Target Distribution ===')
print(y_train.value_counts())
print(f'\nPercentages:')
print((y_train.value_counts(normalize=True) * 100).round(2))

print('\n=== Test Target Distribution ===')
print(y_test.value_counts())
print(f'\nPercentages:')
print((y_test.value_counts(normalize=True) * 100).round(2))
```
